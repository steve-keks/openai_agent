```python
import openai
import dotenv
import os
```

```python
dotenv.load_dotenv()

client = openai.OpenAI(
    api_key=os.getenv("API_KEY"),
)
```

```python
chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "Say this is a test",
        }
    ],
    model="gpt-4o",
)
```
