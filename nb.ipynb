```python
import openai
import dotenv
import os
import json
```

```python
dotenv.load_dotenv()

client = openai.OpenAI(
    api_key=os.getenv("API_KEY"),
)
```

```python

chat_completion = client.chat.completions.create(
    max_completion_tokens=50,
    messages=[
        {
            "role": "user",
            "content": "What is the capital of Finland?,"
        }
    ],
    model="gpt-4o",
)
```

```python
def get_response(prompt, **kwargs):
    chat_completion = client.chat.completions.create(
        max_completion_tokens=100,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        model="gpt-4o",
    )
    return chat_completion.choices[0].message.content
```

```python
print(chat_completion.choices[0].message.content) 
```

```python

response_json = json.dumps(chat_completion, default=lambda o: o.__dict__, indent=4)
print(response_json)
```

```python
from pydantic import BaseModel
from typing import List, Optional, Dict, Any

class Message(BaseModel):
    content: str
    role: str
    tool_calls: Optional[Any] = None
    function_call: Optional[Any] = None
    provider_specific_fields: Optional[Dict[str, Any]] = None
    refusal: Optional[Any] = None

class Choices(BaseModel):
    finish_reason: str
    index: int
    message: Message

class ChatCompletion(BaseModel):
    id: str
    created: int
    model: str
    object: str
    system_fingerprint: Optional[str] = None
    choices: List[Choices]
    _response_ms: float

# Example usage
data = {
    'id': 'chatcmpl-B0wN4jDCxtiNZutXGNVxParIQlqqE',
    'created': 1739564078,
    'model': 'gpt-4-0613',
    'object': 'chat.completion',
    'system_fingerprint': None,
    'choices': [
        {
            'finish_reason': 'length',
            'index': 0,
            'message': {
                'content': "As an artificial intelligence, I don't have feelings",
                'role': 'assistant',
                'tool_calls': None,
                'function_call': None,
                'provider_specific_fields': {'refusal': None},
                'refusal': None
            }
        }
    ],
    '_response_ms': 1146.93
}

chat_completion = ChatCompletion(**data)
print(chat_completion)
```
